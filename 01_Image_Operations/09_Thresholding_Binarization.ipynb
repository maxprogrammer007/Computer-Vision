```python
import numpy as np 
import matplotlib.pyplot as plt
import cv2
```

```python
# defining our imshow function

def imshow(title = 'Image',image = None,size = 10):
    w,h = image.shape[1],image.shape[0]
    aspect_ratio = w/h
    plt.figure(figsize=(size*aspect_ratio,size))
    plt.imshow(cv2.cvtColor(image,cv2.COLOR_BGR2RGB))
    plt.title(title)
    plt.show()
```

```python
# reading the image as grayscale
img = cv2.imread('C:\\Users\\abhin\\OneDrive\\Documents\\GitHub\\Computer-Vision\\data\\aero3.jpg',0)
imshow('Original Image',img)
```

```python
# values below 127 will be converted to 0 and above 127 will be converted to 255

ret , thresh1 = cv2.threshold(img,127,255,cv2.THRESH_BINARY)
imshow('Binary Threshold',thresh1)
```

```python
# values below 127 will be converted to 255 and above 127 will be converted to 0
ret,thresh2 = cv2.threshold(img,127,255,cv2.THRESH_BINARY_INV)
imshow('Binary Inverse Threshold',thresh2)
```

```python
# values above 127  will be truncated to 127 ( the 255 arugment is not used)
ret,thresh3 = cv2.threshold(img,127,255,cv2.THRESH_TRUNC)
imshow('Truncated Threshold',thresh3)
```

```python
# values below 127 go to 0 and above 127 are unchanged

ret,thresh4 = cv2.threshold(img,127,255,cv2.THRESH_TOZERO)
imshow('To Zero Threshold',thresh4)
```

```python
# reverse the above
ret,thresh5 = cv2.threshold(img,127,255,cv2.THRESH_TOZERO_INV)
imshow('To Zero Inverse Threshold',thresh5)
```
